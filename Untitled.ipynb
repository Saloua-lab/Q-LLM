```python
import numpy as np
import matplotlib.pyplot as plt

# -----------------------------
# 1) Remplis tes scores Q-LLM
# -----------------------------
qllm_rmse = 0.0  # TODO: mets ton RMSE Amazon-Books
qllm_mae  = 0.0  # TODO: mets ton MAE  Amazon-Books

# -----------------------------
# Baselines Amazon-Books (du tableau)
# -----------------------------
baselines = {
    # zero-shot LLMs
    "Flan-U-PALM (0-shot)": {"RMSE": 0.9565, "MAE": 0.5569, "AUC": 0.7676},
    "ChatGPT (0-shot)":     {"RMSE": 1.0081, "MAE": 0.8093, "AUC": 0.6778},
    "text-davinci-003 (0-shot)": {"RMSE": 0.8890, "MAE": 0.5442, "AUC": 0.7416},

    # heuristics
    "UAR (heuristic)":      {"RMSE": 0.8527, "MAE": 0.5502, "AUC": 0.8047},

    # supervised recommender (upper bound)
    "Transformer+MLP (sup.)": {"RMSE": 0.8143, "MAE": 0.5541, "AUC": 0.8042},

    # fine-tuned LLM (upper bound)
    "Flan-T5-Base (clf, FT)": {"RMSE": 0.9856, "MAE": 0.4685, "AUC": 0.6292},
    "Flan-T5-XXL (reg, FT)":  {"RMSE": 0.8301, "MAE": 0.5122, "AUC": 0.8312},
}

# -----------------------------
# Figure 1 — RMSE (Amazon)
# Bar plot: text-davinci-003, UAR, Q-LLM + ligne upper bound supervised
# -----------------------------
labels_rmse = ["text-davinci-003 (0-shot)", "UAR (heuristic)", "Q-LLM"]
values_rmse = [
    baselines["text-davinci-003 (0-shot)"]["RMSE"],
    baselines["UAR (heuristic)"]["RMSE"],
    qllm_rmse
]
upper_rmse = baselines["Transformer+MLP (sup.)"]["RMSE"]

x = np.arange(len(labels_rmse))

plt.figure(figsize=(7,4))
plt.bar(x, values_rmse)
plt.axhline(upper_rmse, linestyle="--", linewidth=2, label="Transformer+MLP (supervised)")

plt.xticks(x, labels_rmse, rotation=15, ha="right")
plt.ylabel("RMSE (↓)")
plt.title("Amazon-Books — RMSE comparison")
plt.legend()
plt.tight_layout()
plt.show()


# -----------------------------
# Figure 2 — MAE (Amazon)
# Bar plot: text-davinci-003, UAR, Q-LLM + ligne upper bound fine-tuned
# -----------------------------
labels_mae = ["text-davinci-003 (0-shot)", "UAR (heuristic)", "Q-LLM"]
values_mae = [
    baselines["text-davinci-003 (0-shot)"]["MAE"],
    baselines["UAR (heuristic)"]["MAE"],
    qllm_mae
]
upper_mae = baselines["Flan-T5-Base (clf, FT)"]["MAE"]

x = np.arange(len(labels_mae))

plt.figure(figsize=(7,4))
plt.bar(x, values_mae)
plt.axhline(upper_mae, linestyle="--", linewidth=2, label="Flan-T5-Base (fine-tuned)")

plt.xticks(x, labels_mae, rotation=15, ha="right")
plt.ylabel("MAE (↓)")
plt.title("Amazon-Books — MAE comparison")
plt.legend()
plt.tight_layout()
plt.show()


# -----------------------------
# Figure 3 — Positioning plot (Training cost vs RMSE)
# Scatter: zero-shot LLMs, Q-LLM, supervised, fine-tuned
# X-axis = proxy training cost (0=none, 1=few-shot, 2=supervised, 3=fine-tuned)
# -----------------------------
points = [
    ("Flan-U-PALM (0-shot)", 0, baselines["Flan-U-PALM (0-shot)"]["RMSE"]),
    ("ChatGPT (0-shot)", 0, baselines["ChatGPT (0-shot)"]["RMSE"]),
    ("text-davinci-003 (0-shot)", 0, baselines["text-davinci-003 (0-shot)"]["RMSE"]),
    ("UAR (heuristic)", 0, baselines["UAR (heuristic)"]["RMSE"]),
    ("Q-LLM", 0, qllm_rmse),
    ("Transformer+MLP (sup.)", 2, baselines["Transformer+MLP (sup.)"]["RMSE"]),
    ("Flan-T5-XXL (reg, FT)", 3, baselines["Flan-T5-XXL (reg, FT)"]["RMSE"]),
]

plt.figure(figsize=(7,4))
xs = [p[1] for p in points]
ys = [p[2] for p in points]
plt.scatter(xs, ys)

for name, cx, rmse in points:
    plt.annotate(name, (cx, rmse), textcoords="offset points", xytext=(6,4), ha="left", fontsize=9)

plt.xticks([0,1,2,3], ["No training\n(0-shot/heuristic)", "Few-shot", "Supervised", "Fine-tuned"])
plt.ylabel("RMSE (↓)")
plt.title("Amazon-Books — Performance vs training cost")
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```
