```python
def load_dataset(path: str) -> pd.DataFrame:
    if path.lower().endswith('.csv'):
        return pd.read_csv(path)
    elif path.lower().endswith('.json'):
        return pd.read_json(path)
    else:
        raise ValueError("Unsupported file type. Use CSV or JSON.")
```

```python
def chunk_text(text: str, max_chars: int = 1500) -> List[str]:
    """Simple character-based chunker. Keeps chunks <= max_chars."""
    text = text.replace('\n', ' ').strip()
    if len(text) <= max_chars:
        return [text]
    chunks = []
    start = 0
    while start < len(text):
        end = min(len(text), start + max_chars)
        # try to split on sentence boundary
        if end < len(text):
            # look back for a period
            pb = text.rfind('.', start, end)
            if pb > start:
                end = pb + 1
        chunks.append(text[start:end].strip())
        start = end
    return chunks


# Prompt template
PROMPT = '''Extract entities and relations from the following text. Output strictly valid JSON with two keys: "entities" and "relations".

- "entities" should be a list of objects with: id (unique short string), label (type, e.g., Place, Category, Person), name (surface form), properties (dict of additional attributes, optional).
- "relations" should be a list of objects with: from_id, to_id, relation (verb or relation type), properties (dict, optional).

Example output (for guidance):
{
  "entities": [
    {"id": "e1", "label": "Place", "name": "Joe's Pizza", "properties": {"lat": 40.73061, "lon": -73.935242}},
    {"id": "e2", "label": "Category", "name": "Pizza", "properties": {}}
  ],
  "relations": [
    {"from_id": "e1", "to_id": "e2", "relation": "HAS_CATEGORY", "properties": {"confidence": 0.9}}
  ]
}

Now extract from this text (only return JSON):
'''

# LLM wrapper (API)

class LLMExtractor:
    def __init__(self, model_name: str = HF_MODEL, use_api=False):
        self.use_api = use_api
        self.model_name = model_name
        self.generator = None
        self.tokenizer = None
        if not use_api:
            # load tokenizer + model with Transformers pipeline
            print(f"Loading local model {model_name} (this may take a while)...")
            self.tokenizer = AutoTokenizer.from_pretrained(model_name, trust_remote_code=True)
            self.model = AutoModelForCausalLM.from_pretrained(
                model_name,
                trust_remote_code=True,
                device_map='auto',
                # uncomment if you want to try 8-bit loading and have bitsandbytes installed
                # load_in_8bit=True,
            )
            self.generator = pipeline(
                "text-generation",
                model=self.model,
                tokenizer=self.tokenizer,
                max_length=MAX_TOKENS,
                temperature=TEMPERATURE,
                do_sample=False,
            )
        else:
            if not self.api_key:
                raise ValueError("HF_API_KEY must be set for API usage")
            # use HF inference API wrapper (requests) - simple

    def extract_json(self, text: str) -> Dict[str, Any]:
        prompt = PROMPT + '\n' + text + '\n'
        if self.use_api:
            # simple HF Inference call
            import requests
            headers = {"Authorization": f"Bearer {self.api_key}", "Content-Type": "application/json"}
            body = {"inputs": prompt, "parameters": {"max_new_tokens": MAX_TOKENS, "temperature": TEMPERATURE}}
            url = f"https://api-inference.huggingface.co/models/{self.model_name}"
            resp = requests.post(url, headers=headers, json=body)
            resp.raise_for_status()
            out = resp.json()
            # HF inference returns a list or dict depending on model; extract text
            if isinstance(out, list):
                text_out = out[0].get('generated_text', out[0].get('generated_text', ''))
            elif isinstance(out, dict):
                text_out = out.get('generated_text', '')
            else:
                text_out = str(out)
        else:
            # run local generator
            gen = self.generator(prompt, max_new_tokens=MAX_TOKENS, do_sample=False)[0]['generated_text']
            # sometimes the tokenizer/model echo includes prompt; we try to remove prompt prefix
            if gen.startswith(prompt):
                text_out = gen[len(prompt):].strip()
            else:
                text_out = gen

        # Attempt to find first JSON object in text_out
        try:
            # naive extraction: find first { and last }
            first = text_out.find('{')
            last = text_out.rfind('}')
            json_text = text_out[first:last+1]
            data = json.loads(json_text)
            return data
        except Exception as e:
            print("Failed to parse JSON from LLM output. Returning raw text and error.", e)
            return {"error": str(e), "raw": text_out}


# Neo4j ingestion functions
def get_neo4j_driver(uri: str, user: str, password: str):
    driver = GraphDatabase.driver(uri, auth=(user, password))
    return driver


def upsert_entity(tx, ent):
    # ent expected: {id, label, name, properties}
    props = ent.get('properties', {})
    # create or merge node by unique id
    tx.run(
        "MERGE (n:Entity {uid:$uid})\nSET n.name = $name, n.label = $label, n.props = $props",
        uid=ent['id'], name=ent['name'], label=ent.get('label', 'Entity'), props=props)


def upsert_relation(tx, rel):
    # rel expected: {from_id, to_id, relation, properties}
    tx.run(
        "MATCH (a:Entity {uid:$from}), (b:Entity {uid:$to})\nMERGE (a)-[r:REL {rel_type:$rel_type}]->(b)\nSET r.props = $props",
        from_id=rel['from_id'], to_id=rel['to_id'], rel_type=rel.get('relation', 'RELATED_TO'), props=rel.get('properties', {}))


# Main pipeline

def build_graph_from_dataset(data_path: str, text_column: str = 'description', limit: int = None):
    df = load_dataset(data_path)
    if text_column not in df.columns:
        # try to combine some columns
        df['description'] = df.apply(lambda r: ' '.join([str(r[c]) for c in df.columns if pd.notna(r[c])]), axis=1)
        text_column = 'description'

    if limit is not None:
        df = df.head(limit)

    extractor = LLMExtractor()
    driver = get_neo4j_driver(NEO4J_URI, NEO4J_USER, NEO4J_PASSWORD)

    with driver.session() as session:
        for idx, row in df.iterrows():
            text = str(row[text_column])
            chunks = chunk_text(text, max_chars=1200)
            for i, chunk in enumerate(chunks):
                print(f"Processing record {idx}, chunk {i+1}/{len(chunks)} (len={len(chunk)})")
                out = extractor.extract_json(chunk)
                if 'entities' in out and 'relations' in out:
                    # Upsert entities
                    try:
                        session.write_transaction(lambda tx: [upsert_entity(tx, e) for e in out['entities']])
                        # Upsert relations
                        session.write_transaction(lambda tx: [upsert_relation(tx, r) for r in out['relations']])
                    except Exception as ne:
                        print('Neo4j ingest error:', ne)
                else:
                    print('LLM output did not follow schema:', out)
                # small sleep to avoid throttling
                time.sleep(0.1)

    driver.close()
    print('Done building graph.')
```

```python
if not os.path.exists(DATA_PATH):
    print('Please update DATA_PATH in the script to point to your dataset. Exiting.')
else:
    build_graph_from_dataset(DATA_PATH, text_column='description', limit=100)
```
